```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score, silhouette_score
from mlxtend.frequent_patterns import apriori, association_rules

df_credit = pd.read_excel('data/数据-客户信用记录.xlsx')
df_apply = pd.read_excel('data/申请客户信息.xlsx')
df_consume = pd.read_excel('data/消费历史记录.xlsx')
df_delinquency = pd.read_excel('data/拖欠历史记录.xlsx')

# 合并数据：依据客户号建立整合数据表
df_merged = pd.merge(df_credit, df_apply, on='客户号', how='left')

# 年龄分组
bins_age = [18, 25, 35, 45, 60, np.inf]
labels_age = ['18-25', '25-35', '35-45', '45-60', '60+']
df_merged['年龄_分组'] = pd.cut(df_merged['年龄'], bins=bins_age, labels=labels_age, right=False)

# 特征编码 
categorical_features = ['性别', '教育程度', '车辆情况', '住房情况']
df_final = pd.get_dummies(df_merged, columns=categorical_features, drop_first=True)

# 样本集划分
y = df_final['信用等级']  
X = df_final.drop(['信用等级', '客户号'], axis=1)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 标准化
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 决策树
dt_model = DecisionTreeClassifier(max_depth=5, random_state=42)
dt_model.fit(X_train_scaled, y_train)
dt_pred = dt_model.predict(X_test_scaled)
print("决策树模型准确率：", accuracy_score(y_test, dt_pred))

# 逻辑回归
lr_model = LogisticRegression(solver='liblinear', random_state=42)
lr_model.fit(X_train_scaled, y_train)
print("逻辑回归模型准确率：", lr_model.score(X_test_scaled, y_test))

# SVM
svm = SVC(kernel='linear', random_state=42)
svm.fit(X_train_scaled, y_train)
print("SVM 准确率：", svm.score(X_test_scaled, y_test))

# K-Means 聚类
df_cluster = df_consume[['日均消费金额', '单笔消费最大金额']].dropna()
df_cluster_scaled = StandardScaler().fit_transform(df_cluster)

kmeans = KMeans(n_clusters=2, random_state=42)
labels = kmeans.fit_predict(df_cluster_scaled)
score = silhouette_score(df_cluster_scaled, labels)
print("聚类轮廓系数：", round(score, 3))

# Apriori 关联规则
# 转交易矩阵
df_trans = df_consume.pivot_table(
    index='客户号', columns='消费类别', aggfunc=lambda x: 1, fill_value=0
)

freq = apriori(df_trans, min_support=0.05, use_colnames=True)
rules = association_rules(freq, metric='lift', min_threshold=1.0)

print("关联规则前5条：")
print(rules.sort_values('lift', ascending=False).head())

print("合并后数据示例：")
print(df_final.head())  # 你可截图放报告中


# ----------------- 图2：决策树特征重要性 -----------------
importance = dt_model.feature_importances_
feature_names = X.columns
feat_imp = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importance
}).sort_values(by='Importance', ascending=False).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(data=feat_imp, x='Importance', y='Feature')
plt.title('决策树特征重要性（Top 10）')
plt.tight_layout()
plt.show()


# ----------------- 图3：K-Means 聚类可视化 -----------------
df_cluster['Cluster'] = labels

plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=df_cluster, 
    x='日均消费金额', 
    y='单笔消费最大金额', 
    hue='Cluster'
)
plt.title(f"K-Means聚类结果（Silhouette={round(score,3)}）")
plt.tight_layout()
plt.show()


# ----------------- 图4：聚类轮廓系数显示 -----------------
print(f"KMeans 聚类轮廓系数：{score}")


# ----------------- 图5：Apriori关联规则可视化 -----------------
rules_top = rules.sort_values('lift', ascending=False).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(data=rules_top, x='lift', y='consequents')
plt.title('关联规则提升度 Top10')
plt.tight_layout()
plt.show()

print("关联规则前10条：")
display(rules_top)
```
