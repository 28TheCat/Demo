```python
from sklearn.svm import SVC
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import numpy as np

# 1. 加载数据
file_path = r'E:/Demo/python/银行信用卡欺诈与拖欠行为分析/data/数据-客户信用记录.xlsx'
data = pd.read_excel(file_path)

# 2. 特征选择和目标变量
# 选择相关的特征。'是否存在欺诈' 是我们的目标变量。
features = ['卡类别', '币种代码', '额度', '日均消费金额', '日均次数', '单笔消费最小金额', '单笔消费最大金额', '个人收入_连续']
target = '是否存在欺诈'

X = data[features]
y = data[target]

# 3. 独热编码分类特征
X = pd.get_dummies(X, columns=['卡类别', '币种代码'], drop_first=True)

# 4. 处理缺失值（如果有的话，这里使用中位数填充）
for col in X.columns:
    if X[col].isnull().any():
        X[col] = X[col].fillna(X[col].median())

# 5. 特征缩放
# 对数值特征进行标准化，这对SVM等基于距离的算法很重要
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

# 6. 划分训练集和测试集
# 欺诈数据通常是高度不平衡的，因此我们将使用分层抽样 (stratify=y)
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.3, random_state=42, stratify=y
)

print("数据准备完成。")
# 7. 初始化和训练 SVM 模型
# 使用 'rbf' 核函数
svm_model = SVC(kernel='rbf', random_state=42)
svm_model.fit(X_train, y_train)

print("SVM 模型训练完成。")
# 8. 模型预测
y_pred = svm_model.predict(X_test)

# 9. 模型评估
print("\n## 📈 模型评估结果\n")

# 准确率
accuracy = accuracy_score(y_test, y_pred)
print(f"**总体准确率 (Accuracy):** {accuracy:.4f}\n")

# 混淆矩阵
print("**混淆矩阵:**")
conf_mat = confusion_matrix(y_test, y_pred)
conf_mat_df = pd.DataFrame(
    conf_mat,
    index=['实际：非欺诈 (0)', '实际：欺诈 (1)'],
    columns=['预测：非欺诈 (0)', '预测：欺诈 (1)']
)
print(conf_mat_df)

# 分类报告
print("\n**分类报告:**")
print(classification_report(y_test, y_pred))
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

# 1. 简化数据，只保留分类特征和目标变量
apriori_data = data[['卡类别', '币种代码', '是否存在欺诈']].copy()

# 2. 将数据转换为独热编码格式，为 Apriori 做准备
# 注意：对于 Apriori，我们需要为每个特征的每个值创建一个二元（0/1）列
apriori_encoded = pd.get_dummies(apriori_data, columns=['卡类别', '币种代码'], prefix=['卡类别', '币种代码'])

# 我们将所有列名改为更有意义的格式
apriori_encoded.columns = [f"{col_name}_{val}" if col_name not in ['是否存在欺诈'] else col_name for col_name, val in zip(apriori_encoded.columns, apriori_encoded.mean() > 0)]

# 确保所有列都是布尔类型
apriori_encoded = apriori_encoded.astype(bool)

print("\n## 🔎 Apriori 关联规则挖掘\n")
print("Apriori 数据集准备完成。")
# 3. 运行 Apriori 算法找出频繁项集
# min_support (最小支持度): 至少占总事务的比例
frequent_itemsets = apriori(apriori_encoded, min_support=0.01, use_colnames=True)

# 4. 挖掘关联规则
# metric (度量): 使用 'lift' (提升度) 来衡量规则的强度
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)

# 5. 筛选包含 '是否存在欺诈_True' 的规则，并按提升度排序
fraud_rules = rules[rules['consequents'].apply(lambda x: '是否存在欺诈_True' in str(x))].sort_values(by='lift', ascending=False)

print("\n**前 10 条关联规则（Consequents 包含 '是否存在欺诈_True'）:**")
print(fraud_rules.head(10))
```
